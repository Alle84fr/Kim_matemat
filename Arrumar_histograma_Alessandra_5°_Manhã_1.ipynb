```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
idade = [12,43,14,65,43,55,23,2,32,43,76,87,45,76,87,98,11,9,54,76,87,98,34,23,12,32,4,66,56,23,12,12,32,4,65,77,87,18,65,99,10,12]

#converte a lista (string) para num
idade_np = np.array(idade)
```

```python
contagem_idade=len(idade_np)

media_idade=np.mean(idade_np)

mediana_idade=np.median(idade_np)

variancia=np.var(idade_np)

desvio_padrão=np.std(idade_np)

#menor valor observado
maximo=np.max(idade_np)

# maior valor observado
minimo=np.min(idade_np)

# diferença entre os extremos
amplitude=np.max(idade_np)-np.min(idade_np)

#AIQ - usada para outliers
amplitude_interquartil=np.percentile(idade_np,75)-np.percentile(idade_np,25)

# disperção relativa dos dados em relação à média
coeficiente_variacao=np.std(idade_np)/np.mean(idade_np)

#curtose - distribuição se pontuda ou achatada
mu= np.mean((idade_np - np.mean(idade_np))**4)

curtose=mu/(np.std(idade_np)**4) - 3

```

```python
# 25%
Q1 = np.percentile(idade_np, 25)

# 75%
Q3 = np.percentile(idade_np, 75)

# linha entre horizontal e caixa
limite_inferior = Q1 - 1.5*amplitude_interquartil

limite_superior = Q3 + 1.5*amplitude_interquartil

print(f"Q1 {Q1}")
print(f"Q3 {Q3}")
print(f"Menor valor {limite_inferior}")
print(f"Maior valor {limite_superior}")
```

```python
print(f"Há {contagem_idade} de pessoas")
print(f"A média de  idade é {media_idade:.2f}")
print(f"A mediana de idade é {mediana_idade}")
print(f"A variância de idade é {variancia:.2f}")
print(f"O desvio padrão de idade é {desvio_padrão:.2f}")
print(f"A maior idade é {maximo}")
print(f"A menor idade é {minimo}")
print(f"A amplitude de idade é {amplitude}")
print(f"A amplitude interquartil de idade é {amplitude_interquartil}")
print(f"O coeficiente de variação de idade é {coeficiente_variacao:.2f}")
print(f"A curtose de idade é {curtose:.2f}")
```

.figsize=(10,6) = tamanho do histograma, em podelgadas

idade_np = nome dos dados plotados

kde = True = true para curvas acentuadas

color = cor da barra, que pode ser com #

edgecolor = cor das bordas da barra

.axvline() = axis vertical line (axis = eixo vertical linha)
ob = axhline = axis horizontal line
   = axline = axis linha = linha com inclinação definida, muitas vezes reta

linestyle = "--" = estilo da linha, aqui tracejada

linewidth = 2 = espessura da linha

bin = intervalo =

```python
from matplotlib.lines import lineStyles

# nova janela de gráfio
plt.figure(figsize=(10,6))

# configura a tabela
sns.histplot(idade_np, kde=True)

#__________ fazer barras coloridas
counts, edges, bars = plt.hist(idade_np, bins=[0,20,40,60,80,100])

colors = ["#98f5e1","#8eecf5","#a0c4ff","#bdb2ff","#ffc6ff"]
for bar, color in zip(bars, colors):
    bar.set_facecolor(color)
#___________________________

# linha vertical | relacionada a média | cor da linha | linha traceja | espessura da linha | legenda
plt.axvline(media_idade, color="#7a6c5d", linestyle="--", linewidth=2, label=f"Média: {media_idade:.2f}")

plt.axvline(mediana_idade, color="#f85e00", linestyle="-.", linewidth=2, label=f"Mediana: {mediana_idade}")

plt.axvline(maximo, color="#c75146", linestyle=":", linewidth=2, label=f"máximo: {maximo:.2f}")

plt.axvline(minimo, color="#283618", linestyle=":", linewidth=2, label=f"mínimo: {minimo:.2f}")

plt.title("Histograma Idade Alessandra", fontsize=14, fontweight="bold")

plt.xlabel("Idade")

plt.ylabel("Frequência")

# plt.legend() - queria valor no histograma e não na legenda

plt.text(media_idade+ 1, 0.5, f"Média: {media_idade:.2f}", color="#495057", fontsize=9, rotation=90, va="bottom")

plt.text(mediana_idade- 2, 0.5,  f"Mediana: {mediana_idade}", color="#495057", fontsize=9, rotation=90, va="bottom")

plt.text(maximo+ 1, 0.5,  f"Máx: {maximo}", color="#495057", fontsize=9, rotation=90, va="bottom")

plt.text(minimo+ 1, 0.5,  f"Mín: {minimo}", color="#495057", fontsize=9, rotation=90, va="bottom")


plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(x=idade_np, color="#d5c6e0")

plt.title("Boxplot Idade", fontsize=14, fontweight="bold")

# Quadrado
plt.text(Q1- 0.8, 0.25, f"Q1: {Q1:.1f}", color="#495057", fontsize=9, rotation=90, ha="center")

plt.text(mediana_idade- 0.8, 0.25, f"Q2: {mediana_idade:.1f}", color="#495057", fontsize=9, rotation=90, ha="center")

plt.text(Q3 - 0.8, 0.25, f"Q3: {Q3:.1f}", color="#495057", fontsize=9, rotation=90, ha="center")

# Limites
plt.text(limite_inferior- 0.8, 0.15, f"Limite Infe\n{limite_inferior:.1f}", color="#7b2cbf", fontsize=8, rotation=90, ha="center")

plt.text(limite_superior- 0.8, 0.15, f"Limite Super\n{limite_superior:.1f}", color="#7b2cbf", fontsize=8, rotation=90, ha="center")

# Mínimo e máximo - possíveis oputlier
plt.text(minimo - 0.6, -0.05, f"Mín: {minimo}", color="#283618", fontsize=9, rotation=90, ha="center")

plt.text(maximo- 0.6, -0.05, f"Máx: {maximo}", color="#283618", fontsize=9, rotation=90, ha="center")

plt.xlabel("Idade")

plt.show()
```
