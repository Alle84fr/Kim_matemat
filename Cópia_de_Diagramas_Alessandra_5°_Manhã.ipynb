```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
altura = [1.43,1.56,1.65,1.87,1.11,1.77,1.43,1.43, 1.56, 1.65, 1.87, 1.11, 1.77, 1.43, 1.65, 1.56, 1.70, 1.87, 1.43, 1.60, 1.55, 1.11, 1.77, 1.68, 1.82, 1.43, 1.56, 1.43, 1.65, 1.77, 1.90, 1.52, 1.43, 1.11, 1.65, 1.70, 1.56]
estatura = np.array(altura)
```

```python
tamanho_amostra = len(estatura)

media = np.mean(estatura)

mediana = np.median(estatura)

# variança
var_x = np.var(estatura)

# desvio padrão
s = np.std(estatura)

mini = np.min(estatura)

maxi = np.max(estatura)

amplitude = (maxi) - (mini)

Q1 = np.percentile(estatura, 25)

Q3 = np.percentile(estatura, 75)

# amplitude interquatil
AIQ =  (Q3) - (Q1)

#coeficiente de variação = dados espalhados ou não
cv = s / media

#outlier inferior
outl_inf = Q3 - Q1

#outlier superior
outl_sup = Q3 + Q1

#
mu = np.mean((estatura - media)**4)

curstose = mu / (((s)**4)-3)
```

```python
print(f"""Tamanho da amostra de altura: {tamanho_amostra}
      Média da altura: {media}
      Mediana  da altura / Q2: {mediana}
      Variação da altura: {var_x:.2f}
      Desvio Padrão de altura: {s:.2f}
      Valor mínino: {mini:.2f}
      Valor máximo: {maxi:.2f}
      Amplitude : {amplitude:.2f}
      Outlier inferior: {outl_inf:.2f}
      Outlier ssuperio: {outl_sup:.2f}
      Q1: {Q1}
      Q3: {Q3}
      AIQ/ amplitude interquartil: {AIQ:.2f}
      Coeficiente da variação: {cv:.2f}
      Base para calcular curtose: {mu:.2f}
      Curstose: {curstose:.2f}""")
```

```python
from matplotlib.lines import lineStyles

# nova janela de gráfio
plt.figure(figsize=(14,10))

# cria bins de 10 em 10 até o máximo
bins = np.arange(1.00, 2.01, 0.10)

plt.xticks(np.arange(1.00, 2.01, 0.10))


# histograma com plt.hist
counts, edges, bars = plt.hist(estatura, bins=bins)

# cores diferentes para cada barra
colors = ["#e2e2df","#d2d2cf","#e2cfc4","#f7d9c4","#faedcb","#c9e4de","#c6def1","#dbcdf0","#f2c6de","#f9c6c9"]
for bar, color in zip(bars, colors):
    bar.set_facecolor(color)
#______________________________________

# linha vertical | relacionada a média | cor da linha | linha traceja | espessura da linha | legenda
plt.axvline(media, color="#7a6c5d", linestyle="--", linewidth=2, label=f"Média: {media:.2f}")

plt.axvline(mediana, color="#f85e00", linestyle="-.", linewidth=2, label=f"Mediana: {mediana}")

plt.axvline(maxi, color="#c75146", linestyle=":", linewidth=2, label=f"máximo: {maxi:.2f}")

plt.axvline(mini, color="#283618", linestyle=":", linewidth=2, label=f"mínimo: {mini:.2f}")

plt.title("Histograma Altura Alessandra", fontsize=14, fontweight="bold")

plt.xlabel("Altura")

plt.ylabel("Frequência")

plt.legend()




plt.show()
```

```python
plt.figure(figsize=(45,40))

sns.boxplot(x=estatura, color="#d08c60")

plt.title("Boxplot Altura", fontsize=14, fontweight="bold")

# Quadrado
plt.text(Q1+ 0.01, 0.1, f"Q1: {Q1:.1f}", color="#495057", fontsize=9, rotation=90, ha="center")

plt.text(mediana+ 0.01, 0.1, f"Q2: {mediana:.1f}", color="#495057", fontsize=9, rotation=90, ha="center")

plt.text(Q3+ 0.01, 0.1, f"Q3: {Q3:.1f}", color="#495057", fontsize=9, rotation=90, ha="center")

# Limites
plt.text(outl_inf+ 0.01, 0.1, f"Limite Infe\n{outl_inf:.1f}", color="#7b2cbf", fontsize=8, rotation=90, ha="center")

plt.text(outl_sup+ 0.01, 0.1, f"Limite Super\n{outl_sup:.1f}", color="#7b2cbf", fontsize=8, rotation=90, ha="center")

# Mínimo e máximo - possíveis oputlier
plt.text(mini+ 0.01, 0.1, f"Mín: {mini}", color="#283618", fontsize=9, rotation=90, ha="center")

plt.text(maxi+ 0.01, 0.1, f"Máx: {maxi}", color="#283618", fontsize=9, rotation=90, ha="center")

plt.xlabel("Altura")

plt.show()
```
