```python
import pandas as pd
import numpy as np

# 1. Criando o vetor/DataFrame "sujo"
dados = {
    'ID': [1, 2, 2, 4, 5, 6, 7, 8],
    'Nome': ['Ana', 'Bruno', 'Bruno', 'Carlos', '  DANIELA  ', 'Eduardo', 'Fernanda', 'Gabriel'],
    'Idade': [25, 30, 30, np.nan, 22, 150, 28, np.nan],
    'Salario': ['2500', '3000', '3000', '4000', '5000', 'text_erro', '3500', '4200']
}
df = pd.DataFrame(dados)

# --- INÍCIO DA LIMPEZA ---

# A. Remover Duplicatas (O segundo 'Bruno' e o ID 2 repetido)
df = df.drop_duplicates()

# B. Limpar espaços em branco no texto ('  DANIELA  ' vira 'DANIELA')
df['Nome'] = df['Nome'].str.strip()

# C. Corrigir tipos e tratar erros (Transforma 'text_erro' em NaN para podermos calcular)
df['Salario'] = pd.to_numeric(df['Salario'], errors='coerce')

# D. Tratar valores faltantes (NaN)
# Preenchendo a idade com a mediana (mais seguro que a média)
df['Idade'] = df['Idade'].fillna(df['Idade'].median())
# Preenchendo o salário faltante com a média do grupo
df['Salario'] = df['Salario'].fillna(df['Salario'].mean())

# E. Remover Outliers (O '150' na idade que não serve)
# Usando um filtro simples: apenas idades menores que 100
df = df[df['Idade'] < 100]

print("Dados Limpos e Prontos:")
print(df)
```
